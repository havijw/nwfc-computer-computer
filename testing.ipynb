```python
from computer_computer.models import Course, Student, SubjectArea, Teacher
from computer_computer.solver import ComputerComputerSolver
```

## Sample Data


```python
students = [Student(name=f"Student{i}", is_multilingual_learner=i < 3) for i in range(10)]
teachers = [
    Teacher(name=f"Teacher{i}", subject_area=subject_area)
    for i, subject_area in enumerate(SubjectArea)
]
courses = (
    [
        Course(
            title=f"{teacher.name}'s first period",
            period=1,
            subject_areas=[teacher.subject_area],
            teachers=[teacher],
        )
        for teacher in teachers
    ]
    + [
        Course(
            title=f"{teacher.name}'s second period",
            period=2,
            subject_areas=[teacher.subject_area],
            teachers=[teacher],
        )
        for teacher in teachers
    ]
    + [
        Course(
            title=f"{teacher.name}'s third period",
            period=3,
            subject_areas=[teacher.subject_area],
            teachers=[teacher],
        )
        for teacher in teachers
    ]
)
```

```python
first_choices = {
    student: (
        courses[i % len(teachers)],  # first period
        courses[(i + 1) % len(teachers) + len(teachers)],  # second period
        courses[(i + 2) % len(teachers) + 2 * len(teachers)],  # third period
    )
    for i, student in enumerate(students)
}
second_choices = {
    student: (
        courses[(i + 1) % len(teachers)],  # first period
        courses[(i + 2) % len(teachers) + len(teachers)],  # second period
        courses[(i + 3) % len(teachers) + 2 * len(teachers)],  # third period
    )
    for i, student in enumerate(students)
}
```

## Optimization


```python
solver = ComputerComputerSolver(students=students, courses=courses)
print(len(solver.constraints))
```

```python
first_choices = {
    student: (
        courses[i % len(teachers)],  # first period
        courses[(i + 1) % len(teachers) + len(teachers)],  # second period
        courses[(i + 2) % len(teachers) + 2 * len(teachers)],  # third period
    )
    for i, student in enumerate(students)
}
second_choices = {
    student: (
        courses[(i + 1) % len(teachers)],  # first period
        courses[(i + 2) % len(teachers) + len(teachers)],  # second period
        courses[(i + 3) % len(teachers) + 2 * len(teachers)],  # third period
    )
    for i, student in enumerate(students)
}
course_assignments = solver.get_optimal_course_assignments(first_choices, second_choices, disp=True)
```

```python
for course, students in course_assignments.items():
    print(course.title)
    print("-" * len(course.title))
    for student in students:
        print(student.name)
    print()
```
